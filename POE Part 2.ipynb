# Train the model

```python
#Split the data set into 75% training and 25% testing
from sklearn.model_selection import train_test_split
train, test = train_test_split(df, test_size = 0.25, random_state = 0)
```

```python
train_x = train[predictive_var]
train_y = train['diagnosis']
test_x = test[predictive_var]
test_y = test['diagnosis']
```

```python
#Scale the data - Feature Scalling
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
train_x = sc.fit_transform(train_x)
test_x = sc.fit_transform(test_x)
#test_x
```

```python
#Create a function that will create 3 models (Logistic Regression, Decision Tree and Random Forrest )
def models(train_x, train_y):
    #Logistic Regression
    from sklearn.linear_model import LogisticRegression
    log = LogisticRegression(random_state = 0)
    log.fit(train_x, train_y)

    #Decision Tree
    from sklearn.tree import DecisionTreeClassifier
    dtree = DecisionTreeClassifier(criterion = 'entropy', random_state = 0)
    dtree.fit(train_x, train_y)

    #Random Forest
    from sklearn.ensemble import RandomForestClassifier
    rforest = RandomForestClassifier(n_estimators = 10, criterion = 'entropy', random_state = 0)
    rforest.fit(train_x, train_y)

    #Print the accuracy of the models on the training data
    print('[0] Logistic Regression Accuracy - Training:', log.score(train_x, train_y))
    print('[1] Decision Tree Accuracy - Training:', dtree.score(train_x, train_y))
    print('[2] Random Forest Accuracy - Training:', rforest.score(train_x, train_y))

    #Return the models
    return log, dtree, rforest
    
```

```python
#Run all the models
runmodels = models(train_x, train_y)
```

```python
#Test the model with Testing data
#print(len(test_y), len(test_x))
from sklearn.metrics import classification_report
from sklearn.metrics import accuracy_score
# print( classification_report(test_y, runmodels[0].predict(test_x)))
# print( accuracy_score(test_y, runmodels[0].predict(test_x)))
for i in range (len(runmodels)):
    print('Model ', i)
    print( classification_report(test_y, runmodels[i].predict(test_x)))
    print( accuracy_score(test_y, runmodels[i].predict(test_x)))
    print('______________________________________________________')
```
